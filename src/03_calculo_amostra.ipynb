```python
import math

# Parâmetros convencionais (conforme os slides)
z = 1.96  # Nível de confiança de 95%
p = 0.50  # Proporção esperada de 50%
e = 0.05  # Erro aceitável de 5%

# Fórmula de população alvo (infinita)
n = ((z ** 2) * p * (1 - p)) / (e ** 2)
n = math.ceil(n)

print(f"Tamanho da amostra indicada (população infinita): {n} registros")
```

```python
import pandas as pd

def calcular_amostra_finita(n_infinita, tamanho_populacao):
    # Aplicando o fator de correção
    n_linha = n_infinita / (1 + ((n_infinita - 1) / tamanho_populacao))
    return math.ceil(n_linha)

# Carregar a série (base_unificada.csv) para encontrar o tamanho da população (N)
import os
base_dir = 'bases' if os.path.exists('bases') else '../bases'
df_base = pd.read_csv(os.path.join(base_dir, 'base_unificada.csv'), low_memory=False)
N_total = len(df_base)

print(f"Tamanho da série (N total da base_unificada): {N_total} registros")

amostra_corrigida = calcular_amostra_finita(n, N_total)

print(f"Tamanho da amostra suavizada para essa população finita: {amostra_corrigida} registros")
```
